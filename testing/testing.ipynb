```python
from graphviz import Digraph

dot = Digraph("MultimodalChatAssistant", format="png")
dot.attr(rankdir="LR", size="8", dpi="300")

dot.node("Video", "Video Input (≤10min, 60fps)", shape="box", style="filled", fillcolor="lightblue")
dot.node("Frames", "Frame Extraction / Sampling\n(OpenCV, PyAV)", shape="box", style="filled", fillcolor="lightgrey")
dot.node("Audio", "Audio Extraction & Transcription\n(Whisper, Vosk)", shape="box", style="filled", fillcolor="lightgrey")
dot.node("VisualEvents", "Visual Event Recognition\n(YOLOv8, SlowFast, VideoMAE)", shape="box", style="filled", fillcolor="lightyellow")
dot.node("TranscriptAnalysis", "Transcript Analysis & Key Moment Detection\n(TextRank, BERT, LLM)", shape="box", style="filled", fillcolor="lightyellow")
dot.node("Fusion", "Multimodal Fusion\n(Combine Visual + Audio Insights)", shape="box", style="filled", fillcolor="orange")
dot.node("Summarization", "Summarization Engine\n(LLM/VLM e.g., LLaVA, BLIP-2)", shape="box", style="filled", fillcolor="lightgreen")
dot.node("VectorDB", "Knowledge Store / Vector DB\n(FAISS, Chroma)", shape="box", style="filled", fillcolor="lightpink")
dot.node("Chat", "Multi-turn Chat Assistant\n(FastAPI + LLM, Context Memory)", shape="box", style="filled", fillcolor="lightblue")

dot.edges([("Video", "Frames"), ("Video", "Audio")])
dot.edge("Frames", "VisualEvents")
dot.edge("Audio", "TranscriptAnalysis")
dot.edge("VisualEvents", "Fusion")
dot.edge("TranscriptAnalysis", "Fusion")
dot.edge("Fusion", "Summarization")
dot.edge("Summarization", "VectorDB")
dot.edge("VectorDB", "Chat")
dot.edge("Summarization", "Chat")

dot.render("multimodal_chat_architecture_hd", format="png", cleanup=True)
```
